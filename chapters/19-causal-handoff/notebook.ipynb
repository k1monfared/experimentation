# Chapter 19: Causal handoff

Mostly conceptual. The compact demonstration: selection bias in observational comparisons.

```python
import numpy as np
from expkit.plot.style import apply_style
apply_style()
```

```python
rng = np.random.default_rng(190)
n = 5000
propensity = rng.normal(0, 1, n)  # latent confounder: affects both treatment and outcome
treated = rng.random(n) < 1 / (1 + np.exp(-propensity))
outcome = propensity + 0.30*treated + rng.normal(0, 0.5, n)
naive = outcome[treated].mean() - outcome[~treated].mean()
print(f'true effect (ATE) = +0.30')
print(f'naive observational diff (SDO) = {naive:+.3f}')
print(f'  Rubin decomposition: SDO = ATE + selection bias + HTE bias')
print(f'  Treatment effect is constant here (no heterogeneity), so HTE bias = 0.')
print(f'  Selection bias accounts for the full gap: {naive - 0.30:+.3f}')
```
