# Chapter 9: Wine and the small print

```python
import numpy as np
from expkit.sim.user_segments import SegmentSpec, segmented_binary
from expkit.plot.style import apply_style
apply_style()
```

## Loop A: external validity

```python
specs = [
    SegmentSpec('22yo college students', 0.10, 0.40, 0.05),
    SegmentSpec('40yo under stress', 0.20, 0.55, -0.02),
    SegmentSpec('65+ healthy adults', 0.20, 0.30, 0.01),
    SegmentSpec('pregnant women', 0.05, 0.45, -0.10),
    SegmentSpec('teens', 0.10, 0.20, -0.05),
    SegmentSpec('middle-aged sedentary', 0.35, 0.50, 0.00),
]
pop = segmented_binary(20000, specs, seed=901)
for s in specs:
    diff = pop[s.name]['treatment'].mean() - pop[s.name]['control'].mean()
    print(f'{s.name:<28} diff = {diff:+.3f}  (true = {s.treatment_lift:+.3f})')
```

## Loop D: tradeoff weights

```python
target_effect, harm_effect = 0.30, -0.40
for w in [0.3, 0.5, 0.57, 0.7, 0.9]:
    u = w * target_effect + (1 - w) * harm_effect
    print(f'weight on target = {w:.2f}  ->  utility = {u:+.3f}  ->  decision = {"ship" if u > 0 else "do not ship"}')
```
