## Loop B: clickbait dynamics

```python
days = np.arange(0, 30)
click = 0.10 * np.exp(-days/7)
retention = -0.05 * (1 - np.exp(-days/14))
for d in [1, 7, 14, 30]:
    print(f'day {d:>2}: click lift = {100*click[d-1]:+.2f}%   retention lift = {100*retention[d-1]:+.2f}%')
```

## Loop C: aggregate vs cohorts

```python
fractions = [0.10, 0.30, 0.60]
lifts = [-0.02, 0.03, 0.06]
for name, f, l in zip(['most engaged 10%', 'moderately 30%', 'long tail 60%'], fractions, lifts):
    print(f'{name:<20} fraction={f}  lift={l:+.3f}')
print(f'aggregate lift = {sum(f*l for f,l in zip(fractions, lifts)):+.3f}')
```

## Loop D: guardrails fire false alarms

```python
rng = np.random.default_rng(0)
n_metrics = 20; n_runs = 5000
naive = bonf = 0
for _ in range(n_runs):
    p = rng.uniform(0, 1, n_metrics)
    if any(x < 0.05 for x in p): naive += 1
    if any(x < 0.05/n_metrics for x in p): bonf += 1
print(f'family-wise type-I rate, naive    : {naive/n_runs:.3f}')
print(f'family-wise type-I rate, Bonferroni: {bonf/n_runs:.3f}')
```
