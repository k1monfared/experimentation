```python
import numpy as np
from expkit.metrics.quality import predictivity, relative_noise
from expkit.plot.style import apply_style
apply_style()
```

## Predictivity of short-term vs long-term

```python
rng = np.random.default_rng(0)
n = 200
truth = rng.normal(0, 0.02, size=n)
short_term = truth*0.6 + rng.normal(0, 0.03, n)
long_term = truth*0.9 + rng.normal(0, 0.01, n)
print(predictivity(short_term, long_term))
print('relative noise on short-term metric:', round(relative_noise(short_term), 3))
print('relative noise on long-term metric:', round(relative_noise(long_term), 3))
```
