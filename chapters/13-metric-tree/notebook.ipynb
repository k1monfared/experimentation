```python
import numpy as np
from scipy.stats import spearmanr
from expkit.metrics.quality import predictivity, signal_to_noise
from expkit.plot.style import apply_style
apply_style()
```

## Predictivity of short-term vs long-term

```python
rng = np.random.default_rng(0)
n = 200
truth = rng.normal(0, 0.02, size=n)
short_term = truth*0.6 + rng.normal(0, 0.03, n)
long_term = truth*0.9 + rng.normal(0, 0.01, n)

# Predictivity: Pearson r with a 95% bootstrap CI, plus Spearman rho.
# Pearson assumes linearity; on real (saturating, threshold) metric pairs,
# report Spearman alongside.
pred = predictivity(short_term, long_term, n_boot=1000, seed=0)
rho, _ = spearmanr(short_term, long_term)
print(pred)
print(f"Spearman rho: {rho:.3f}")

# Noise summary: these arrays are zero-centered (truth ~ N(0, 0.02)),
# so the coefficient of variation explodes near the origin and is
# misleading. Use signal_to_noise against a meaningful reference scale
# (here: an MDE of 0.01 in the same units as the effect).
mde = 0.01
print('short-term std / MDE:', round(signal_to_noise(short_term, reference_scale=mde), 3))
print('long-term std / MDE:', round(signal_to_noise(long_term, reference_scale=mde), 3))
print('short-term std (raw):', round(signal_to_noise(short_term), 4))
print('long-term std (raw):', round(signal_to_noise(long_term), 4))
```
