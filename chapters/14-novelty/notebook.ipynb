# Chapter 14: Novelty effects

```python
import numpy as np
from expkit.sim.novelty import novelty_event_log
from expkit.novelty.measure import by_calendar_day, by_days_since_exposure, by_exposure_count
from expkit.plot.style import apply_style
apply_style()
```

```python
df = novelty_event_log(n_users=1500, days=30, base_rate=0.10, initial_lift=0.06, half_life_days=5.0, seed=140)
print('total rows:', len(df))
print(df.head())
```

## Three views

```python
cal = by_calendar_day(df)
sin = by_days_since_exposure(df)
cnt = by_exposure_count(df)
print('--- calendar day ---'); print(cal.head().to_string(index=False))
print('\n--- days since exposure ---'); print(sin.head().to_string(index=False))
print('\n--- exposure count ---'); print(cnt.head().to_string(index=False))
```
