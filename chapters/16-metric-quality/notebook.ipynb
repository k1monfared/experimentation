# Chapter 16: Metric quality

```python
import numpy as np
from scipy import stats
from expkit.metrics.variance import cuped, variance_reduction_ratio
from expkit.metrics.quality import predictivity, stability_aa, aa_calibration
from expkit.plot.style import apply_style
apply_style()
```

## CUPED

```python
rng = np.random.default_rng(0)
n = 4000
pre = rng.normal(10, 3, n)
y = 0.7*pre + rng.normal(0, 1, n)
y_adj, theta = cuped(y, pre)
print(f'theta = {theta:.3f}, variance reduction = {100*variance_reduction_ratio(y, y_adj):.1f}%')
```

## A/A stability

```python
rng = np.random.default_rng(0)
effects = []
p_values = []
for _ in range(2000):
    c = rng.normal(0, 1, 500); t = rng.normal(0, 1, 500)
    effects.append(t.mean() - c.mean())
    p_values.append(stats.ttest_ind(t, c, equal_var=False).pvalue)
print('shape check (stability_aa):', stability_aa(np.array(effects)))
print('calibration check (aa_calibration):', aa_calibration(np.array(p_values), alpha=0.05))
```

## Predictivity with bootstrap CI

```python
rng = np.random.default_rng(7)
n = 200
truth = rng.normal(0, 0.02, size=n)
short_good = truth * 0.8 + rng.normal(0, 0.02, n)
long_term = truth * 0.95 + rng.normal(0, 0.005, n)
print(predictivity(short_good, long_term, n_boot=1000, seed=0))
```
