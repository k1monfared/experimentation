# Chapter 16: Metric quality

```python
import numpy as np
from expkit.metrics.variance import cuped, variance_reduction_ratio
from expkit.metrics.quality import predictivity, stability_aa
from expkit.plot.style import apply_style
apply_style()
```

## CUPED

```python
rng = np.random.default_rng(0)
n = 4000
pre = rng.normal(10, 3, n)
y = 0.7*pre + rng.normal(0, 1, n)
y_adj, theta = cuped(y, pre)
print(f'theta = {theta:.3f}, variance reduction = {100*variance_reduction_ratio(y, y_adj):.1f}%')
```

## A/A stability

```python
rng = np.random.default_rng(0)
effects = []
for _ in range(2000):
    c = rng.normal(0, 1, 500); t = rng.normal(0, 1, 500)
    effects.append(t.mean() - c.mean())
print(stability_aa(np.array(effects)))
```
