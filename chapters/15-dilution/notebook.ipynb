# Chapter 15: Dilution

```python
import numpy as np
from expkit.sim.dilution import settings_page_experiment
from expkit.metrics.delta import delta_two_arm
from expkit.plot.style import apply_style
apply_style()
```

## ITT vs per-protocol

```python
df = settings_page_experiment(20000, visit_prob=0.12, base_rate=0.20, treatment_lift_among_visitors=0.10, seed=150)
yt = df[df['arm']=='treatment']['outcome'].values
yc = df[df['arm']=='control']['outcome'].values
itt = yt.mean() - yc.mean()
se_itt = np.sqrt(yt.var(ddof=1)/len(yt) + yc.var(ddof=1)/len(yc))
sub = df[df['visited']]
yvt = sub[sub['arm']=='treatment']['outcome'].values
yvc = sub[sub['arm']=='control']['outcome'].values
pp = yvt.mean() - yvc.mean()
se_pp = np.sqrt(yvt.var(ddof=1)/len(yvt) + yvc.var(ddof=1)/len(yvc))
print(f'ITT effect = {itt:+.4f}, SE = {se_itt:.4f}  (expected ~ 0.12 * 0.10 = 0.012)')
print(f'per-protocol effect = {pp:+.4f}, SE = {se_pp:.4f}  (true among-visitors = 0.10)')
print(f'A single 20k draw can land near zero or even negative on ITT: |expected/SE| ~ {0.012/se_itt:.2f}')
```

## Delta method on a ratio metric

```python
rng = np.random.default_rng(0)
n = 3000
sess_c = rng.poisson(8, n); rev_c = rng.normal(2.0*sess_c, 4)
sess_t = rng.poisson(8, n); rev_t = rng.normal(2.05*sess_t, 4)
diff, se, z = delta_two_arm(rev_t, sess_t, rev_c, sess_c)
print(f'delta method on revenue/session: diff = {diff:+.4f}, se = {se:.4f}, z = {z:.2f}')
```
