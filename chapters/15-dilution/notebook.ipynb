# Chapter 15: Dilution

```python
import numpy as np
from expkit.sim.dilution import settings_page_experiment
from expkit.metrics.delta import delta_two_arm
from expkit.plot.style import apply_style
apply_style()
```

## ITT vs per-protocol

```python
df = settings_page_experiment(20000, visit_prob=0.12, base_rate=0.20, treatment_lift_among_visitors=0.10, seed=150)
itt = df[df['arm']=='treatment']['outcome'].mean() - df[df['arm']=='control']['outcome'].mean()
sub = df[df['visited']]
pp = sub[sub['arm']=='treatment']['outcome'].mean() - sub[sub['arm']=='control']['outcome'].mean()
print(f'ITT effect = {itt:+.4f} (visit rate ~ 0.12, true among-visitors = 0.10)')
print(f'per-protocol effect = {pp:+.4f}')
```

## Delta method on a ratio metric

```python
rng = np.random.default_rng(0)
n = 3000
sess_c = rng.poisson(8, n); rev_c = rng.normal(2.0*sess_c, 4)
sess_t = rng.poisson(8, n); rev_t = rng.normal(2.05*sess_t, 4)
diff, se, z = delta_two_arm(rev_t, sess_t, rev_c, sess_c)
print(f'delta method on revenue/session: diff = {diff:+.4f}, se = {se:.4f}, z = {z:.2f}')
```
